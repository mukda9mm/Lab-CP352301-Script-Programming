# lab4_1.py - Student Management System (List Operations)

```python
# 1. สร้าง List ว่าง
student_names = []

# 2. เพิ่มรายชื่อนักเรียน (Add Students) 3-5 คน
num = int(input("ต้องการเพิ่มนักเรียนกี่คน (3-5): "))
while num < 3 or num > 5:
    num = int(input("กรุณากรอกจำนวนระหว่าง 3-5: "))

for i in range(num):
    name = input(f"กรอกชื่อนักเรียนคนที่ {i+1}: ")
    student_names.append(name)
print("รายชื่อปัจจุบัน:", student_names)

# 3. ค้นหารายชื่อนักเรียน (Find a Student)
search_name = input("ต้องการค้นหาชื่อใคร: ")
if search_name in student_names:
    print(f"พบ {search_name} อยู่ที่ index {student_names.index(search_name)}")
else:
    print(f"ไม่พบ {search_name} ในรายชื่อ")

# 4. ลบรายชื่อนักเรียน (Remove a Student)
remove_name = input("ต้องการลบชื่อใคร: ")
if remove_name in student_names:
    student_names.remove(remove_name)
    print(f"ลบ {remove_name} เรียบร้อยแล้ว")
else:
    print(f"ไม่พบ {remove_name} ในรายชื่อ จึงลบไม่ได้")
print("รายชื่อหลังอัปเดต:", student_names)

# 5. จัดเรียงรายชื่อ (Sort Students)
student_names.sort()
print("รายชื่อหลังเรียงลำดับ:", student_names)

# 6. นับจำนวนนักเรียน (Count Students)
print("จำนวนนักเรียนทั้งหมด:", len(student_names), "คน")
```

# Lab 4.1 Challenge: ระบบเมนู

```python
student_names = []

while True:
    print("\n===== ระบบจัดการรายชื่อนักเรียน =====")
    print("1. เพิ่ม")
    print("2. ลบ")
    print("3. ค้นหา")
    print("4. เรียงลำดับ")
    print("5. ออกจากโปรแกรม")
    choice = input("เลือกเมนู (1-5): ")

    if choice == "1":
        name = input("กรอกชื่อนักเรียนที่ต้องการเพิ่ม: ")
        student_names.append(name)
        print(f"เพิ่ม {name} แล้ว")
        print("รายชื่อปัจจุบัน:", student_names)

    elif choice == "2":
        name = input("กรอกชื่อนักเรียนที่ต้องการลบ: ")
        if name in student_names:
            student_names.remove(name)
            print(f"ลบ {name} แล้ว")
        else:
            print(f"ไม่พบ {name} ในรายชื่อ")
        print("รายชื่อปัจจุบัน:", student_names)

    elif choice == "3":
        name = input("กรอกชื่อนักเรียนที่ต้องการค้นหา: ")
        if name in student_names:
            print(f"พบ {name} อยู่ที่ index {student_names.index(name)}")
        else:
            print(f"ไม่พบ {name} ในรายชื่อ")

    elif choice == "4":
        student_names.sort()
        print("รายชื่อหลังเรียงลำดับ:", student_names)

    elif choice == "5":
        print("ออกจากโปรแกรม จำนวนนักเรียนทั้งหมด:", len(student_names), "คน")
        break

    else:
        print("กรุณาเลือกเมนู 1-5 เท่านั้น")
```

# lab4_2.py - Geometric Calculations (Tuples & Immutability)

```python
import math

# 1. กำหนดพิกัดจุดเป็น Tuple
point1 = (3, 4)
point2 = (6, 8)

# เข้าถึงค่าพิกัด
print(f"Point 1: x = {point1[0]}, y = {point1[1]}")
print(f"Point 2: x = {point2[0]}, y = {point2[1]}")

# 2. ทดลองแก้ไขค่า (Immutability)
# point1[0] = 5
# ผลลัพธ์: TypeError: 'tuple' object does not support item assignment
# เหตุผล: Tuple เป็นชนิดข้อมูลแบบ immutable สร้างแล้วจะเปลี่ยนแปลง เพิ่ม หรือลบสมาชิกไม่ได้
# ถ้าต้องการค่าใหม่ต้องสร้าง tuple ใหม่ เช่น point1 = (5, 4)

# คำนวณระยะห่าง: sqrt((x2 - x1)^2 + (y2 - y1)^2)
distance = math.sqrt((point2[0] - point1[0]) ** 2 + (point2[1] - point1[1]) ** 2)
print("ระยะห่างระหว่าง point1 และ point2 =", distance)
```

# Lab 4.2 Challenge: รับค่าด้วย input

```python
import math

# รับค่าแล้วแปลงเป็น float ก่อนนำไปคำนวณ (Type Conversion)
x1 = float(input("กรอกพิกัด x ของจุดที่ 1: "))
y1 = float(input("กรอกพิกัด y ของจุดที่ 1: "))
x2 = float(input("กรอกพิกัด x ของจุดที่ 2: "))
y2 = float(input("กรอกพิกัด y ของจุดที่ 2: "))

point1 = (x1, y1)
point2 = (x2, y2)
print(f"Point 1: x = {point1[0]}, y = {point1[1]}")
print(f"Point 2: x = {point2[0]}, y = {point2[1]}")

distance = math.sqrt((point2[0] - point1[0]) ** 2 + (point2[1] - point1[1]) ** 2)
print(f"ระยะห่างระหว่างสองจุด = {distance:.2f}")
```

# Activity 4.1: List vs. Tuple: Scenario Matching

| ข้อ | สถานการณ์ | ควรใช้ | เหตุผล |
|---|---|---|---|
| 1 | อุณหภูมิประจำวันตลอดหนึ่งสัปดาห์ | **List** | ต้องเพิ่มค่าเข้าไปทีละวันขณะบันทึก และอาจต้องแก้ไขค่าที่บันทึกผิด |
| 2 | ค่าสี RGB เช่น (255, 0, 0) | **Tuple** | ค่าสีคงที่ มีสมาชิก 3 ค่าตายตัว ถ้าแก้ไขก็จะกลายเป็นสีอื่น |
| 3 | สินค้าในรถเข็นที่เพิ่มหรือลบได้ | **List** | ต้องใช้ append() และ remove() เพิ่ม/ลบสินค้าได้ตลอด |
| 4 | ชื่อวันในหนึ่งสัปดาห์ (จันทร์–อาทิตย์) | **Tuple** | ข้อมูลคงที่ ลำดับและจำนวน 7 วันไม่เปลี่ยน ป้องกันการแก้ไขโดยไม่ตั้งใจ |
| 5 | คะแนนเก็บที่อาจแก้ไขภายหลัง | **List** | คะแนนเพิ่มขึ้นเมื่อส่งงานใหม่และอาจถูกแก้ไข จึงต้องเปลี่ยนแปลงได้ |
| 6 | พิกัด GPS ของแลนด์มาร์ก | **Tuple** | สถานที่อยู่กับที่ พิกัด (ละติจูด, ลองจิจูด) จึงคงที่ ไม่ควรถูกเปลี่ยน |

**สรุป:** ใช้ List เมื่อข้อมูลต้องเพิ่ม ลบ หรือแก้ไขได้ (mutable) และใช้ Tuple เมื่อข้อมูลคงที่ ไม่ควรถูกแก้ไข (immutable)

# Activity 4.2: To-Do List

```python
# 1. สร้าง List ว่างเริ่มต้น
todo_list = []

# 2. ใช้ while loop แสดงเมนู
while True:
    print("\n===== To-Do List =====")
    print("1. เพิ่มรายการงาน (Add a task)")
    print("2. ดูรายการงานทั้งหมด (View tasks)")
    print("3. ทำเครื่องหมายว่างานเสร็จสิ้น (Mark task as done)")
    print("4. ออกจากโปรแกรม (Exit)")
    # 3. รับค่าเมนูจากผู้ใช้
    choice = input("เลือกเมนู (1-4): ")

    if choice == "1":
        task = input("กรอกงานที่ต้องการเพิ่ม: ")
        todo_list.append(task)
        print(f"เพิ่มงาน '{task}' แล้ว")

    elif choice == "2":
        if len(todo_list) == 0:
            print("ยังไม่มีงานในรายการ")
        else:
            print(f"งานทั้งหมด {len(todo_list)} รายการ:")
            for i in range(len(todo_list)):
                print(f"  {i + 1}. {todo_list[i]}")

    elif choice == "3":
        if len(todo_list) == 0:
            print("ยังไม่มีงานในรายการ")
        else:
            for i in range(len(todo_list)):
                print(f"  {i + 1}. {todo_list[i]}")
            item = input("กรอกหมายเลขงาน หรือชื่องานที่เสร็จแล้ว: ")
            if item.isdigit():
                # ลบตามตำแหน่ง (หมายเลขเริ่มที่ 1 จึงลบ 1 ให้เป็น index)
                index = int(item) - 1
                if 0 <= index < len(todo_list):
                    done = todo_list.pop(index)
                    print(f"งาน '{done}' เสร็จสิ้นแล้ว")
                else:
                    print("หมายเลขงานไม่ถูกต้อง")
            elif item in todo_list:
                # ลบตามชื่องาน
                todo_list.remove(item)
                print(f"งาน '{item}' เสร็จสิ้นแล้ว")
            else:
                print(f"ไม่พบงาน '{item}' ในรายการ")

    elif choice == "4":
        print(f"ออกจากโปรแกรม เหลืองานที่ยังไม่เสร็จ {len(todo_list)} รายการ")
        break

    else:
        print("กรุณาเลือกเมนู 1-4 เท่านั้น")
```
