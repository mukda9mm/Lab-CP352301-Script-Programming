# Lab 6.1: Simple Calculator Functions
ฟังก์ชัน `add`, `subtract`, `multiply`, `divide` พร้อม docstring, เมนูด้วย while loop, ดัก division by zero

**Challenge:** `power(base, exponent=2)` ใช้ default parameter (เมนูข้อ 5)

```python
# ============================================================
# Lab 6.1: Simple Calculator Functions
# ============================================================

def add(a, b):
    """คืนค่าผลบวกของ a และ b"""
    return a + b


def subtract(a, b):
    """คืนค่าผลลบของ a ลบด้วย b"""
    return a - b


def multiply(a, b):
    """คืนค่าผลคูณของ a และ b"""
    return a * b


def divide(a, b):
    """คืนค่าผลหารของ a หารด้วย b
    ถ้า b เป็น 0 จะคืนข้อความแจ้ง Error แทน เพื่อป้องกัน ZeroDivisionError
    """
    if b == 0:
        return "Error: Division by zero"
    return a / b


# ---------- Challenge: Default Parameter ----------
def power(base, exponent=2):
    """คืนค่า base ยกกำลัง exponent
    ถ้าไม่ส่ง exponent มา จะใช้ค่า default = 2 (ยกกำลังสอง)
    """
    return base ** exponent


# ---------- Menu-driven program ----------
while True:
    print("\n===== Simple Calculator =====")
    print("1. Add (+)")
    print("2. Subtract (-)")
    print("3. Multiply (*)")
    print("4. Divide (/)")
    print("5. Power (**)  [Challenge]")
    print("6. Exit")
    choice = input("Enter choice (1-6): ")

    if choice == "6":
        print("Exiting calculator. Goodbye!")
        break

    if choice in ("1", "2", "3", "4"):
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))

        if choice == "1":
            result = add(num1, num2)
            print(f"{num1} + {num2} = {result}")
        elif choice == "2":
            result = subtract(num1, num2)
            print(f"{num1} - {num2} = {result}")
        elif choice == "3":
            result = multiply(num1, num2)
            print(f"{num1} * {num2} = {result}")
        elif choice == "4":
            result = divide(num1, num2)
            print(f"{num1} / {num2} = {result}")

    elif choice == "5":
        base = float(input("Enter base: "))
        exp_input = input("Enter exponent (กด Enter เพื่อใช้ค่า default = 2): ")
        if exp_input == "":
            result = power(base)               # ใช้ default parameter
            print(f"{base} ** 2 (default) = {result}")
        else:
            exponent = float(exp_input)
            result = power(base, exponent)     # ส่ง exponent เอง
            print(f"{base} ** {exponent} = {result}")

    else:
        print("Invalid choice! Please enter a number from 1 to 6.")
```

# Lab 6.2: Custom Module and Standard Library Usage
## Part 1: Create a Custom Module (`my_utils.py`)
ใช้คำสั่ง `%%writefile` สร้างไฟล์ `my_utils.py` ใน Colab

```python
%%writefile my_utils.py
# my_utils.py - Custom Module
"""โมดูลรวมฟังก์ชันอรรถประโยชน์ (utility functions) สำหรับ Lab 6.2"""


def greet(name):
    """พิมพ์ข้อความทักทายตามชื่อที่รับเข้ามา"""
    print(f"Hello, {name}!")


def is_prime(number):
    """คืนค่า True ถ้า number เป็นจำนวนเฉพาะ (prime) ไม่เช่นนั้นคืนค่า False"""
    if number < 2:
        return False
    # ตรวจตัวหารตั้งแต่ 2 ถึงรากที่สองของ number ก็พอ
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False
    return True
```

## Part 2: Use the Custom Module and Standard Modules (`main_program.py`)

```python
# main_program.py - ใช้งาน Custom Module และ Standard Modules
import my_utils      # โมดูลที่เราสร้างเอง
import math          # Standard Library
import random        # Standard Library

# 1. เรียกใช้ greet()
my_utils.greet("Alice")

# 2. ทดสอบ is_prime() กับหลาย ๆ ตัวเลข
for n in [7, 10, 13, 1, 25, 97]:
    print(f"is_prime({n}) = {my_utils.is_prime(n)}")

# 3. ใช้ math.sqrt() หารากที่สอง
number = 144
print(f"\nSquare root of {number} = {math.sqrt(number)}")

# 4. ใช้ random.randint() สุ่มเลข 1-100
random_number = random.randint(1, 100)
print(f"Random number (1-100): {random_number}")
print(f"Is {random_number} prime? {my_utils.is_prime(random_number)}")
```

**งานที่เลือก:** Week 2 – Number Classifier (ตรวจว่าตัวเลขเป็นบวก/ลบ/ศูนย์ และคู่/คี่)

**ปัญหาของโค้ดเดิม:** ตรรกะ if-elif ถูกเขียนรวมไว้ในโปรแกรมหลัก ถ้าต้องตรวจหลายตัวเลขต้องคัดลอกโค้ดชุดเดิมซ้ำ (ขัดกับหลัก DRY) และอ่าน/แก้ไขยาก

**สิ่งที่ Refactor:**
| ฟังก์ชัน | หน้าที่ |
|---|---|
| `is_positive_negative_zero(num)` | คืนค่าว่าเป็นบวก ลบ หรือศูนย์ |
| `is_even_odd(num)` | คืนค่าว่าเป็นเลขคู่หรือคี่ |
| `classify_number(num)` | เรียกสองฟังก์ชันข้างบนแล้วพิมพ์ผลรวม |

**ผลที่ได้:** แต่ละฟังก์ชันทำหน้าที่เดียว ทดสอบแยกได้ เรียกใช้ซ้ำกับตัวเลขกี่ตัวก็ได้ด้วยบรรทัดเดียว และถ้าต้องแก้ตรรกะก็แก้ที่ฟังก์ชันเดียว

### Before: โค้ดเดิม (ไม่มีฟังก์ชัน)

```python
# ---------- BEFORE: โค้ดเดิม Week 2 Number Classifier (ไม่มีฟังก์ชัน) ----------
# ต้องเขียน if-elif ซ้ำทุกครั้งที่ต้องการตรวจตัวเลขใหม่
num = int(input("Enter a number: "))

if num > 0:
    print(f"{num} is Positive")
elif num < 0:
    print(f"{num} is Negative")
else:
    print(f"{num} is Zero")

if num % 2 == 0:
    print(f"{num} is Even")
else:
    print(f"{num} is Odd")
```

### After: Refactor เป็นฟังก์ชัน

```python
# ---------- AFTER: Refactor ให้เป็นฟังก์ชัน ----------

def is_positive_negative_zero(num):
    """คืนค่า 'Positive', 'Negative' หรือ 'Zero' ตามค่าของ num"""
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    else:
        return "Zero"


def is_even_odd(num):
    """คืนค่า 'Even' ถ้า num เป็นเลขคู่ และ 'Odd' ถ้าเป็นเลขคี่"""
    if num % 2 == 0:
        return "Even"
    return "Odd"


def classify_number(num):
    """รวมผลการจำแนกตัวเลขทั้งหมด แล้วพิมพ์ออกมาในบรรทัดเดียว"""
    sign = is_positive_negative_zero(num)
    parity = is_even_odd(num)
    print(f"{num:>4} -> {sign}, {parity}")


# ใช้งาน: เรียกฟังก์ชันซ้ำได้กับหลายตัวเลขโดยไม่ต้องเขียน if-elif ใหม่
print("--- ทดสอบกับหลายตัวเลข ---")
for n in [15, -8, 0, 42, -3]:
    classify_number(n)

print("\n--- รับค่าจากผู้ใช้ (พิมพ์ q เพื่อออก) ---")
while True:
    text = input("Enter a number (or q to quit): ")
    if text.lower() == "q":
        print("Goodbye!")
        break
    if text.lstrip("-").isdigit():
        classify_number(int(text))
    else:
        print("กรุณากรอกจำนวนเต็มเท่านั้น")
```

## ส่วนที่ 2: สำรวจ Python Standard Library
อ้างอิง: [docs.python.org/3/library](https://docs.python.org/3/library/)

1. **`datetime`** – จัดการวันที่และเวลา เช่น คำนวณจำนวนวันถึงวันสอบ หรือบันทึกเวลาที่ทำรายการ
2. **`collections`** – โครงสร้างข้อมูลเพิ่มเติม เช่น `Counter` นับความถี่ได้ในบรรทัดเดียว (ทำแบบ Word Frequency Counter ของ Week 5 ได้สั้นลงมาก)
3. **`os`** – ทำงานกับระบบปฏิบัติการ เช่น ดูไฟล์ในโฟลเดอร์ ตรวจว่าไฟล์มีอยู่ไหม (ใช้เช็กว่ามี `my_utils.py` แล้วหรือยัง)

ตัวอย่างการใช้งานจริงอยู่ในเซลล์โค้ดด้านล่าง

```python
# ตัวอย่างการใช้งาน Standard Library ที่สำรวจ
import datetime
from collections import Counter
import os

# 1) datetime: นับวันถอยหลังถึงวันสอบ
today = datetime.date.today()
exam_day = datetime.date(today.year, 12, 15)
if exam_day < today:
    exam_day = datetime.date(today.year + 1, 12, 15)
print("Today:", today)
print("Days until exam (15 Dec):", (exam_day - today).days, "days")

# 2) collections.Counter: นับความถี่ของคำในบรรทัดเดียว
sentence = "python is fun and python is easy"
word_counts = Counter(sentence.split())
print("\nWord counts:", dict(word_counts))
print("Most common word:", word_counts.most_common(1))

# 3) os: ตรวจสอบไฟล์ในโฟลเดอร์ปัจจุบัน
print("\nDoes my_utils.py exist?", os.path.exists("my_utils.py"))
print("Python files here:", [f for f in os.listdir(".") if f.endswith(".py")])
```
